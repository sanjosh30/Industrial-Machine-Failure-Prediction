```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv("ai4i2020.csv")
```

```python
# 1. Thermal Strain (Temperature gap)
# Heavy metals mein thermal expansion aur heat dissipation issues ko capture karne ke liye
df['Thermal_Strain'] = df['Process temperature [K]'] - df['Air temperature [K]']

# 2. Mechanical Power (Load & Stress)
# Tool wear aur mechanical fatigue/creep ka direct indicator
df['Power'] = df['Torque [Nm]'] * df['Rotational speed [rpm]']

# Check karne ke liye naye columns print karo
#print(df[['Process temperature [K]', 'Air temperature [K]', 'Thermal_Strain', 'Torque [Nm]', 'Rotational speed [rpm]', 'Power']].head())
```

```python
df=df.drop(columns=["UDI","Product ID","TWF","HDF","PWF","OSF","RNF"])
```

```python
df = pd.get_dummies(df, columns=['Type'], dtype=int)
```

```python
x=df.drop("Machine failure",axis=1)
y=df["Machine failure"]
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
from sklearn.compose import ColumnTransformer
# 1. Sirf un columns ke naam likho jinhe scale karna hai (Continuous variables)
num_cols = ['Air temperature [K]', 'Process temperature [K]', 
            'Rotational speed [rpm]', 'Torque [Nm]', 'Tool wear [min]',"Thermal_Strain","Power"]

# 2. ColumnTransformer define karo
# 'passthrough' ka matlab hai ki jo columns num_cols mein nahi hain (jaise Type_L, Type_M, Type_H), unhe bina chhede aage pass kar do
preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), num_cols) # NUM sirf ek naam diya gya h scale krne ko
    ],
    remainder='passthrough'
)

# 3. Ab fit_transform sirf numericals par apply hoga aur baaki waise hi jud jayenge
x_train_scale = preprocessor.fit_transform(x_train)

# Test data ko transform karna
x_test_scale = preprocessor.transform(x_test)
```

```python
import torch
import torch.nn as nn
```

```python
x_train_scale_tensor=torch.tensor(x_train_scale,dtype=torch.float32)
x_test_scale_tensor=torch.tensor(x_test_scale,dtype=torch.float32)

y_train_tensor=torch.tensor(y_train.values,dtype=torch.float32).view(-1,1)
y_test_tensor=torch.tensor(y_test.values,dtype=torch.float32).view(-1,1)
```

```python
from torch.utils.data import TensorDataset, DataLoader
```

```python
train_dataset=TensorDataset(x_train_scale_tensor,y_train_tensor)
test_dataset=TensorDataset(x_test_scale_tensor,y_test_tensor)
```

```python
train_loader=DataLoader(train_dataset,batch_size=32,shuffle=True)
test_loader=DataLoader(test_dataset,batch_size=32)
```

# Define model

```python
class Ann(nn.Module):

    def __init__(self):
        super(Ann,self).__init__()

        self.model=nn.Sequential(
            nn.Linear(10,64),
            nn.ReLU(),
            nn.Dropout(0.2),
            # 1st

            nn.Linear(64,32),
            
            nn.ReLU(),
            nn.Dropout(0.2),
            # 2nd 

            nn.Linear(32,1)
        )
        
    def forward(self,x):
        return self.model(x)
        
```

```python
import torch.optim as optim
```

```python
count_1=(y_train==1).sum()
count_0=(y_train==0).sum()
pos_weight= count_0/count_1
print(f"Calculated pos_weight: {pos_weight.item():.2f}")
pos_weight = torch.tensor([32.3], dtype=torch.float32)
```

```python
a_model=Ann()
crietrion=nn.BCEWithLogitsLoss(pos_weight=pos_weight)
optimizer=optim.Adam(a_model.parameters())
```

```python
for i in range (100):
    a_model.train()
    for xb, xy in train_loader:
        optimizer.zero_grad()
        outputs=a_model(xb)
        loss=crietrion(outputs,xy)
        loss.backward()
        optimizer.step()
```

```python
from sklearn.metrics import confusion_matrix, classification_report
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Model ko 'Evaluation Mode' mein set karna (is se Dropout layers off ho jati hain)
a_model.eval()

# 2. Gradient calculation band karna (Testing ke time backprop nahi hota, isse memory bachti hai)
with torch.no_grad():
    # Test data ko model mein pass karna
    # (Agar tumne X_test ko tensor mein convert kiya tha toh uska naam X_test_tensor hoga)
    raw_outputs = a_model(x_test_scale_tensor)
    
    # 3. Raw outputs (logits) ko 0 ya 1 mein convert karna
    # Agar value > 0 hai, toh 1 (Failure), warna 0 (Normal)
    y_pred = (raw_outputs > 0.0).float()

# 4. PyTorch tensors ko numpy arrays mein convert karna (scikit-learn ke liye zaroori hai)
y_true_np = y_test_tensor.numpy()
y_pred_np = y_pred.numpy()

# ==========================================
# 5. Metrics Calculate aur Print Karna
# ==========================================

print("--- CLASSIFICATION REPORT ---")
# Yeh tumhe Recall, Precision, aur F1-Score sab ek sath nikal kar de dega
print(classification_report(y_true_np, y_pred_np))

print("\n--- CONFUSION MATRIX ---")
cm = confusion_matrix(y_true_np, y_pred_np)
print(cm)

# 6. (Optional) Confusion Matrix ko visually plot karna
plt.figure(figsize=(6, 4))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', 
            xticklabels=['Normal (0)', 'Failure (1)'], 
            yticklabels=['Normal (0)', 'Failure (1)'])
plt.ylabel('Actual Label')
plt.xlabel('Predicted Label')
plt.title('Predictive Maintenance Confusion Matrix')
plt.show()
```
